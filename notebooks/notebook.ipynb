# Carga y diagnóstico — Ames Housing



## 1. Tamaño del conjunto


```python
from pathlib import Path
import pandas as pd

BASE = Path.cwd()
ROOT = BASE if (BASE / "data").exists() else BASE.parent
ARCHIVO = ROOT / "data" / "raw" / "AmesHousing.csv"
df = pd.read_csv(ARCHIVO)

print(f"Filas: {df.shape[0]:,}")
print(f"Columnas: {df.shape[1]}")
```

## 2. Tipo, nulos y valores distintos por variable


```python
diagnostico = pd.DataFrame({
    "Variable": df.columns,
    "Tipo de dato": df.dtypes.astype(str).values,
    "Nulos": df.isna().sum().values,
    "Valores distintos": df.nunique(dropna=True).values,
})

con_nulos = diagnostico.loc[diagnostico["Nulos"] > 0, ["Variable", "Nulos"]]
print(f"Total de valores nulos: {int(diagnostico['Nulos'].sum()):,}")
print(f"Columnas con nulos: {len(con_nulos)} de {len(diagnostico)}")
print("Nulos por columna:")
print(con_nulos.to_string(index=False))

display(diagnostico)
```
