```python
import pandas as pd
import numpy as np

# 1. Cargar el dataset desde la URL de tu GitHub
url = 'AmesHousing.csv'
df = pd.read_csv(url)

# 2. Resumen general de dimensiones
filas, columnas = df.shape
print("=== RESUMEN GENERAL ===")
print(f"Total de Filas (Observaciones): {filas}")
print(f"Total de Columnas (Variables): {columnas}\n")

# 3. Diagnóstico completo de variables
diagnostico = pd.DataFrame({
    'Tipo_Dato': df.dtypes,
    'Cantidad_Nulos': df.isnull().sum(),
    'Porcentaje_Nulos (%)': (df.isnull().sum() / filas) * 100,
    'Valores_Distintos': df.nunique()
})

print("=== DIAGNÓSTICO DE VARIABLES (PRIMERAS 15) ===")
print(diagnostico.head(15))

# 4. Resumen para la tabla del reporte IEEE (Nombres corregidos con espacios)
variables_clave = ['SalePrice', 'Lot Area', 'Overall Qual', 'Neighborhood', 'Year Built', 'Mas Vnr Area', 'Pool QC', 'Gr Liv Area', 'Full Bath']

# Filtrar solo las variables que existen realmente en el dataset
variables_existentes = [v for v in variables_clave if v in diagnostico.index]
diccionario_tabla = diagnostico.loc[variables_existentes]

print("\n=== RESUMEN PARA TABLA IEEE ===")
print(diccionario_tabla)
```
